```python
# differrent types of blurring
import cv2
import matplotlib.pyplot as plt
import numpy as np
img=cv2.imread('tree.jpg')#image read 
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
# 1.box blur(averaging filter)
box=cv2.blur(img,(31,31))
# gaussian blur
gauss=cv2.GaussianBlur(img,(31,31),0)#dimensions of the kernal must be of any odd no
# median blur:
med=cv2.medianBlur(img,31)
plt.subplot(2,2,1),plt.imshow(img),plt.axis('off'),plt.title('source')
plt.subplot(2,2,2),plt.imshow(box),plt.axis('off'),plt.title('box')
plt.subplot(2,2,3),plt.imshow(gauss),plt.axis('off'),plt.title('gaussian')
plt.subplot(2,2,4),plt.imshow(gauss),plt.axis('off'),plt.title('median')
```

```python
# 4.motion blur
# i,horizontal blur    ii. vertical blur
import cv2
import matplotlib.pyplot as plt
import numpy as np

img=cv2.imread('tree.jpg')
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
size=90

# horizontal blur
k1=np.zeros((size,size))
k1[int(size/2),:]=np.ones(size)
k1=k1/size
k1_h=cv2.filter2D(img,-1,k1)

# vertical blur
k2=np.zeros((size,size))
k2[:,int(size/2)]=np.ones(size)
k2=k2/size
k2_v=cv2.filter2D(img,-1,k2)

plt.subplot(1,3,1),plt.imshow(img),plt.title('source image'),plt.axis('off')
plt.subplot(1,3,2),plt.imshow(k1_h),plt.title('horizontal image'),plt.axis('off')
plt.subplot(1,3,3),plt.imshow(k2_v),plt.title('vertical image'),plt.axis('off')
```

```python
# Edge Detection
# 1.sobel2.laplacian3.canny
import cv2
import matplotlib.pyplot as plt

# using sobel(horizontal,vertical)=combine are used
img=cv2.imread('cartoon.png')
img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)# use BGR2GRAY not use BGR2RGB

# Sobel(source_image,ddepth,x,y,kernal_size) by default kernal_size=3
sobel_x=cv2.Sobel(img,cv2.CV_64F,1,0,20)
sobel_y=cv2.Sobel(img,cv2.CV_64F,0,1,20)
sobel_combine=cv2.magnitude(sobel_x,sobel_y)

plt.subplot(1,4,1),plt.imshow(img),plt.title('source image'),plt.axis('off')
plt.subplot(1,4,2),plt.imshow(sobel_x,cmap='plasma'),plt.title('sobel_x '),plt.axis('off')
plt.subplot(1,4,3),plt.imshow(sobel_y,cmap='hot'),plt.title('sobel_y'),plt.axis('off')
plt.subplot(1,4,4),plt.imshow(sobel_combine,cmap='pink'),plt.title('sobel_combine'),plt.axis('off')
```

```python
# Edge detection
# 2.laplacian 
import cv2
import matplotlib.pyplot as plt
img=cv2.imread('cartoon.png')
img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

# gaussian blur
gaus=cv2.GaussianBlur(img,(5,5),0)

# laplacian(src_imd,ddepth,k_size)
lap=cv2.Laplacian(gaus,cv2.CV_64F)#odd numbers
lap=cv2.convertScaleAbs(lap)#used to adjust brightness and contrast.generaly to an 8-bit unsignd integer

plt.subplot(1,2,1),plt.imshow(img),plt.title('source image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(lap,cmap='jet'),plt.title('laplacian'),plt.axis('off')
```

```python
# Erosiaon and Dilation
import cv2
import matplotlib.pyplot as plt
import numpy as np
# jo source image background black and image white to image black mean undhu kari dese
# erosian che te white color ne decrease karse jyare dilation che te white color ne incrase karse
img=cv2.imread('23.png')
img=cv2.bitwise_not(img)

img1=cv2.imread('digit.jpg')
img1=cv2.bitwise_not(img1)

# uint8=unsigned integer 8 bit
kernal=np.ones((3,3),np.uint8)
kernal1=np.ones((3,3),np.uint8)

erosian=cv2.erode(img,kernal,iterations=8)
dilation=cv2.dilate(img,kernal,iterations=8)

erosian1=cv2.erode(img1,kernal1,iterations=8)
dilation1=cv2.dilate(img1,kernal1,iterations=18)

plt.subplot(3,3,1),plt.imshow(img,cmap='gray'),plt.axis('off'),plt.title('Source img')
plt.subplot(3,3,2),plt.imshow(erosian,cmap='gray'),plt.axis('off'),plt.title('erosian')
plt.subplot(3,3,3),plt.imshow(dilation,cmap='gray'),plt.axis('off'),plt.title('dilation')

plt.subplot(3,3,4),plt.imshow(img1,cmap='gray'),plt.axis('off'),plt.title('Source img1')
plt.subplot(3,3,5),plt.imshow(erosian1,cmap='gray'),plt.axis('off'),plt.title('erosian1')
plt.subplot(3,3,6),plt.imshow(dilation1,cmap='gray'),plt.axis('off'),plt.title('dilation1')

```

```python
# contrast effect on color image using GHE
# GHE are use for all image ma eksarkhi lighting apse 
import cv2
import matplotlib.pyplot as plt

img=cv2.imread('r1.jpg')

# Convert to YCrCb
ycrcb=cv2.cvtColor(img,cv2.COLOR_BGR2YCrCb)
y,cr,cb=cv2.split(ycrcb)

# applying Global Equalization Histogram
y_eq=cv2.equalizeHist(y)
ycrcb_eq=cv2.merge([y_eq,cr,cb])
global_eq=cv2.cvtColor(ycrcb_eq,cv2.COLOR_YCrCb2BGR)

# figsize image size mate che keti rakhvi 
plt.figure(figsize=(5,6))
#aa jo image colorfull hoy toj convert karvani che
# cv2.cvtColor(img,cv2.COLOR_BGR2RGB
#jo gray image hoy to convert karvani jarur nathi
plt.subplot(2,2,1),plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)),plt.title('Source image'),plt.axis('off')
plt.subplot(2,2,2),plt.imshow(cv2.cvtColor(global_eq,cv2.COLOR_BGR2RGB)),plt.title('Constract using GHE'),plt.axis('off')
```

```python
# contrast effect on color image using CLAHE
# CLAHE are use for jya jetli jarur hoy tetli j lighting ave
import cv2
import matplotlib.pyplot as plt

img=cv2.imread('r1.jpg')

# Convert to YCrCb
ycrcb=cv2.cvtColor(img,cv2.COLOR_BGR2YCrCb)
y,cr,cb=cv2.split(ycrcb)

# applying CLAHE
clahe=cv2.createCLAHE(clipLimit=2.0,tileGridSize=(8,8))
y_clahe=clahe.apply(y)
ycrcb=cv2.merge([y_clahe,cr,cb])
clahe_eq=cv2.cvtColor(ycrcb,cv2.COLOR_YCrCb2BGR)

# figsize image size mate che keti rakhvi 
plt.figure(figsize=(5,6))
# cv2.cvtColor(img,cv2.COLOR_BGR2RGB
plt.subplot(2,2,1),plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)),plt.title('Source image'),plt.axis('off')
plt.subplot(2,2,2),plt.imshow(cv2.cvtColor(clahe_eq,cv2.COLOR_BGR2RGB)),plt.title('Constract using clahe'),plt.axis('off')
```

```python
# Advanced image filters
# 1.Smoothing effect using Bilsteral
import cv2
import matplotlib.pyplot as plt

img=cv2.imread('r1.jpg')
bilateral = cv2.bilateralFilter(img,d=1,sigmaColor=15,sigmaSpace=15)

plt.subplot(1,2,1),plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)),plt.title('Source image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(cv2.cvtColor(bilateral,cv2.COLOR_BGR2RGB)),plt.title('Bilateral Filter'),plt.axis('off')
```

```python
# Advanced image filters
# 2.Advanced Sharpening using Unsharp
import  cv2
import matplotlib.pyplot as plt

img=cv2.imread('r1.jpg')
gauss=cv2.GaussianBlur(img,(9,9),10)

# Unsharp
unsharp=cv2.addWeighted(img,1.5,gauss,-0.5,0) #(source_img * 1.5)+ (blured_img - 0.5)

plt.subplot(1,2,1),plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)),plt.title('Source image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(cv2.cvtColor(unsharp,cv2.COLOR_BGR2RGB)),plt.title('Unsharp Filter'),plt.axis('off')
```

```python
# Advanced image filters
# 3.edge detection using Scharr

import cv2
import matplotlib.pyplot as plt

img=cv2.imread('lion.jpg',0)

# Scharr
scharrx=cv2.Scharr(img,cv2.CV_64F,1,0)
scharry=cv2.Scharr(img,cv2.CV_64F,0,1)

scharrx=cv2.convertScaleAbs(scharrx)
scharry=cv2.convertScaleAbs(scharry)


# combine
scharr=cv2.addWeighted(scharrx,0.10,scharry,0.1,0)

plt.subplot(1,2,1),plt.imshow(img,cmap='gray'),plt.title('Source image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(scharr,cmap='gray'),plt.title('Scharr'),plt.axis('off')
```

```python
# Object detection
# Face from the image
import cv2

# Load the pre-trained face detector  
face_cascade=cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

# read the image
img=cv2.imread('tmkoc.jpg')
gray_images=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY) #haar Cascade works on single channel color

# detect the face
faces=face_cascade.detectMultiScale(gray_images,scaleFactor=1.1,minNeighbors=3)
print(faces)

# Draw a rectangle around the fase(s)
for(x,y,h,w) in faces:
    cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),3)
cv2.imshow('Face Detection',img)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Object detection
# Face from the camera
import cv2


# Load the pre-trained face detector
face_cascade=cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

# open webcam
cap=cv2.VideoCapture(0) #0 means default camera, 1 means external camera

while True:
    ret,frame=cap.read()
    if not ret:
        print('cannot detact camera')
        break # if the camera is not available

    gray_images=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY) #haar Cascade works on single channel color

# detect the face
    faces=face_cascade.detectMultiScale(gray_images,scaleFactor=1.1,minNeighbors=3)
    print(faces)

# Draw a rectangle around the fase(s)
    for(x,y,h,w) in faces:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),3)
    cv2.imshow('Face Detection from camera:',frame)
    if cv2.waitKey(1)==27:
        break
cap.release()
cv2.destroyAllWindows()
```
