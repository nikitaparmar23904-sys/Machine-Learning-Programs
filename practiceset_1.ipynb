```python
# 1.Write a Python program to apply Erosion and Dilation on a binary image.
import cv2
import matplotlib.pyplot as plt
import numpy as np
#erosion ma white ghate
# dilation ma white vadhe
img=cv2.imread('digit.jpg')
img=cv2.bitwise_not(img)

kernal=np.ones((3,3),np.uint8)

erosian=cv2.erode(img,kernal,iterations=18)
dilation=cv2.dilate(img,kernal,iterations=7)

plt.subplot(1,3,1),plt.imshow(img,cmap='gray'),plt.axis('off'),plt.title('source image')
plt.subplot(1,3,2),plt.imshow(erosian,cmap='gray'),plt.axis('off'),plt.title('erosian')
plt.subplot(1,3,3),plt.imshow(dilation,cmap='gray'),plt.axis('off'),plt.title('dilation')
```

```python
# 2.Write a program to extract object boundaries using (concept of erosion and dilation).

import cv2
import matplotlib.pyplot as plt
import numpy as np
#erosion ma white ghate
# dilation ma white vadhe
img=cv2.imread('digit.jpg')
img=cv2.bitwise_not(img)

kernal=np.ones((3,3),np.uint8)

erosian=cv2.erode(img,kernal,iterations=18)
dilation=cv2.dilate(img,kernal,iterations=7)

#boundary khali joti hoy to source_img - erosian karo to avi gase
img1=img-erosian

plt.subplot(1,4,1),plt.imshow(img,cmap='gray'),plt.axis('off'),plt.title('source image')
plt.subplot(1,4,2),plt.imshow(erosian,cmap='gray'),plt.axis('off'),plt.title('erosian')
plt.subplot(1,4,3),plt.imshow(dilation,cmap='gray'),plt.axis('off'),plt.title('dilation')
plt.subplot(1,4,4),plt.imshow(img1,cmap='gray'),plt.axis('off'),plt.title('boundaries')
```

```python
# 3. Write a program to Load a grayscale image, Apply Global Histogram Equalization and Display
# before and after results.

import cv2
import matplotlib.pyplot as plt
import numpy as np

img=cv2.imread('f.jpg')

#convert YCrCb
ycrcb=cv2.cvtColor(img,cv2.COLOR_BGR2YCrCb)
y,cr,cb=cv2.split(ycrcb)

#appling GHE
y_eq=cv2.equalizeHist(y)
ycrcb_eq=cv2.merge([y_eq,cr,cb])
global_eq=cv2.cvtColor(ycrcb_eq,cv2.COLOR_YCrCb2BGR)

plt.figure(figsize=(5,8))

plt.subplot(2,2,1),plt.imshow(img),plt.axis('off'),plt.title('Source Image')
plt.subplot(2,2,2),plt.imshow(global_eq),plt.axis('off'),plt.title('GHE')
```

```python
# 7.
#Apply CLAHE to the Gray Scale image.

import cv2
import matplotlib.pyplot as plt
import numpy as np

img=cv2.imread('f.jpg')

#convert YCrCb
ycrcb=cv2.cvtColor(img,cv2.COLOR_BGR2YCrCb)
y,cr,cb=cv2.split(ycrcb)

#appling clahe
clahe=cv2.createCLAHE(clipLimit=2.0,tileGridSize=(8,8))
y_clahe=clahe.apply(y)
ycrcb=cv2.merge([y_clahe,cr,cb])
clahe_eq=cv2.cvtColor(ycrcb,cv2.COLOR_YCrCb2BGR)

# figsize image size mate che keti rakhvi 
plt.figure(figsize=(5,6))
# cv2.cvtColor(img,cv2.COLOR_BGR2RGB
plt.subplot(2,2,1),plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)),plt.title('Source image'),plt.axis('off')
plt.subplot(2,2,2),plt.imshow(cv2.cvtColor(clahe_eq,cv2.COLOR_BGR2RGB)),plt.title('Constract using clahe'),plt.axis('off')
```

```python
# 4.
#Apply equalization on RGB, compare the result with applying equalization on YCrCb.
# contrast effect on color image using GHE
# GHE are use for all image ma eksarkhi lighting apse 
import cv2
import matplotlib.pyplot as plt

img=cv2.imread('r1.jpg')
# convert to RGB
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
r, g, b = cv2.split(img)

r_eq = cv2.equalizeHist(r)
g_eq = cv2.equalizeHist(g)
b_eq = cv2.equalizeHist(b)

rgb_eq = cv2.merge((r_eq, g_eq, b_eq))



# Convert to YCrCb
ycrcb=cv2.cvtColor(img,cv2.COLOR_BGR2YCrCb)
y,cr,cb=cv2.split(ycrcb)

# applying Global Equalization Histogram
y_eq=cv2.equalizeHist(y)
ycrcb_eq=cv2.merge([y_eq,cr,cb])
global_eq=cv2.cvtColor(ycrcb_eq,cv2.COLOR_YCrCb2BGR)

# figsize image size mate che keti rakhvi 
plt.figure(figsize=(8,8))
#aa jo image colorfull hoy toj convert karvani che
# cv2.cvtColor(img,cv2.COLOR_BGR2RGB
#jo gray image hoy to convert karvani jarur nathi
plt.subplot(2,3,1),plt.imshow(img),plt.title('Source image'),plt.axis('off')
plt.subplot(2,3,2),plt.imshow(global_eq),plt.title('Constract using GHE'),plt.axis('off')
plt.subplot(2,3,3),plt.imshow(rgb_eq),plt.title('rgb'),plt.axis('off')
```

```python
# 5.
# Display source image, image after applying Sobel and image after applying Scharr.
import cv2
import matplotlib.pyplot as plt

img=cv2.imread('lion.jpg')
img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

# sobel
sobel_x=cv2.Sobel(img,cv2.CV_64F,1,0,20)
sobel_y=cv2.Sobel(img,cv2.CV_64F,0,1,20)
sobel_co=cv2.magnitude(sobel_x,sobel_y)

# Scharr
scharrx=cv2.Scharr(img,cv2.CV_64F,1,0)
scharry=cv2.Scharr(img,cv2.CV_64F,0,1)

scharrx=cv2.convertScaleAbs(scharrx)
scharry=cv2.convertScaleAbs(scharry)


# combine
scharr=cv2.addWeighted(scharrx,0.10,scharry,0.1,0)

plt.subplot(1,3,1),plt.imshow(img),plt.title('Source image'),plt.axis('off')
plt.subplot(1,3,2),plt.imshow(scharr),plt.title('Scharr'),plt.axis('off')

plt.subplot(1,3,3),plt.imshow(sobel_co),plt.title('sobel_combine'),plt.axis('off')
```

```python
# 10. Detect face from the image, put a bounding box around the face.

# Object detection
# Face from the image
import cv2

# Load the pre-trained face detector  
face_cascade=cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

# read the image
img=cv2.imread('tmkoc.jpg')
gray_images=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY) #haar Cascade works on single channel color

# detect the face
faces=face_cascade.detectMultiScale(gray_images,scaleFactor=1.1,minNeighbors=3)
print(faces)

# Draw a rectangle around the fase(s)
for(x,y,h,w) in faces:
    cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),3)
cv2.imshow('Face Detection',img)
cv2.waitKey()
cv2.destroyAllWindows()
```
