```python
# Sentence tokenization
from nltk.tokenize import sent_tokenize

# Taking the text
text='Do you know, I study in Atmiya? Located on kalwad road ,hiiii. hbjgbj'

# Sentence Tokinizing 
sent_tok=sent_tokenize(text)
print('Sentence tokeniser:',sent_tok)
```

```python
# Word tokenization
from nltk.tokenize import word_tokenize

# Taking the text
text="I love NLP. It's fun, isn't it?"

# Word Tokinizing 
word_tok=word_tokenize(text)
print('Word tokeniser:',word_tok)
```

```python
# Punctuation tokenization
from nltk.tokenize import wordpunct_tokenize

# Taking the text
text="I love NLP. It's fun, isn't it?"

# Word Tokinizing 
wordpunct_tok=wordpunct_tokenize(text)
print('Word Punctuation tokeniser:',wordpunct_tok)
```

```python
# Steming (Porter Stemmer, Snowball Stemmer, Lancaster Stemmer)
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer

# Creating Stemmer Objects
porter=PorterStemmer()
snowball=SnowballStemmer('english')
lancaster=LancasterStemmer()

words=['student','playing','played','student','studies','studying','happiness']

print('Porter Stemmer')
for w in words:
    print(w,'->',porter.stem(w))
print('**********')

print('Snowball Stemmer')
for w in words:
    print(w,'->',snowball.stem(w))
print('***********')

print('Lancaster Stemmer')
for w in words:
    print(w,'->',lancaster.stem(w))
```

```python
# Lemmatization
import stanza

# Initializing the NLP pipeline
nlp=stanza.Pipeline(lng='en', processors='tokenize,mwt,pos,lemma')

# Input text
text='The children are learning with their toys an games are going.'

# Processing the text
doc=nlp(text)

# Iterating  over the result
for sentence in doc.sentences:
    for word in sentence.words:
        print(word.text, '-->', word.lemma)
```

```python
# Synthesizing the music
# pip install sounddevice
import numpy as np
import sounddevice as sd

sr=88100 # Sampling rate. Higher the sampling rate, better would be the quality

duration=5.0 #Duration of the note.The music will last for 2 seconds.
freq=440 #Frequency=no. of cycles (vibrations) per second, it is measured in Hertz. 

# Generating the time axis
t=np.linspace(0,duration, int(sr*duration), endpoint=False)

# Creates the time array from 0 to 2 seconds
# Each point representes a tiny  slice of time. Example: 0.0001,0.0002, 0.0003 and so on
# endpoint=False, ensures the sequence stops just before duration

# Generating sine wave
y=0.5*np.sin(2*np.pi*freq*t)
# np.sin(...) generates the sin wave
# 2*np.pi = converts frequency into radians (since sin in maths uses radians) 
# freq*t= no of cycles per second

sd.play(y,sr) #plays the generated signal y using the sampling rate st.
sd.wait()
```

```python
# Identifying the person
# pip install librosa
# pip install sounddevice

import numpy as np 
import librosa
from sklearn.neighbors import KNeighborsClassifier
import sounddevice as sd
import soundfile as sf

# Extracting the features
def extract_features(file_path):
    y,sr = librosa.load(file_path,duration=3) #load the audio file into y, sr= sampling rate, duration is 3 seconds
    mfcc=librosa.feature.mfcc(y=y,sr=sr,n_mfcc=13) #MFCC capture the tone, pitch and timbre of a voice,
    # Extract 13 features for each short frame of audio.
    # y is the parameter, the other y is value of amplitude,
    # sr is the  parameter , the other sr is value of sample rate.
    return np.mean(mfcc.T, axis=0) #It averages the MFCCs over the time giving one  fixed length feature

# Training the data
train_files={
    'a':r"D:\mca_3\ml\sample.wav",
    'b':r"D:\mca_3\ml\sample.mp3",
}
a,b=[],[]
for speaker, file in train_files.items():
    features=extract_features(file)
    a.append(features)
    b.append(speaker)

# Training the classifier
model=KNeighborsClassifier(n_neighbors=1)
model.fit(a,b)

# Recording the audio for testing purpose
duration=3 #for 3 seconds
print('Please spack something: ')

# Recording using microphone
sd.default.device=6 
recording= sd.rec(int(duration*44100), samplerate= 44100, channels=1)
# rec is the function of sounddevice, 3 secx4410 sampling rate, 1 means single channel audio
sd.wait() #it will wait till the recording is over

# Saving the audio file
test_file='test.wav'
sf.write(test_file, recording,44100)

# Prediction
test_features=extract_features(test_file)
predicted_speaker=model.predict([test_features])
print('Based on the live audio, the person spoke is: ',predicted_speaker)
```
