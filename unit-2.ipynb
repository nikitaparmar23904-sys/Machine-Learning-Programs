```python
# pre-processing
# 1.Mean Removal
# always sum of all mean answer is 0
import numpy as np
a=np.array([[2],[4],[6],[8],[10]])

print('The Original Array:',  a)

# calculating the  mean
mean=np.mean(a)
print('The Mean of the numbers: ',mean)

# Substract the mean from each number of the array
centered = a-mean
print('Data after mean removal: ' ,centered)
```

```python
# pre-processing
# 2.Scaling
# scaled always between 0 to 1 
import numpy as np
a=np.array([[10],[20],[30],[40],[50]])
print('The Original data:',a)

# Finding the minimum and maximum value
min_val=np.min(a)
print(min_val)
max_val=np.max(a)
print(max_val)

scaled=(a-min_val)/(max_val-min_val)
print('Scaled data:',scaled)
```

```python
# pre-processing
# 3.Normalization
import numpy as np
from sklearn import preprocessing

data=np.array([[4,4],[7,2],[10,12]])
print('Original data:',data)

# L1
normalized_l1=preprocessing.normalize(data,norm='l1')
print('After L1 Normalization:',normalized_l1)

# L2
normalized_l2=preprocessing.normalize(data,norm='l2')
print('After L1 Normalization:',normalized_l2)
```

```python
# pre-processing
# 4.Binarization (0 or 1)
import numpy as np
from sklearn import preprocessing

data=np.array([[5],[7],[12],[15],[25]])
print('Original data:',data)
print('*'*20)

# threshold=7 means value grater than of 7 than 1 and equal to or less than 7 so 0 
binarized=preprocessing.Binarizer(threshold=7).transform(data) #Applies binarizer to each element
print(binarized)
```

```python
# pre-processing
# 5.Label Encoding 
# follow alphabetical order 
from sklearn import preprocessing
# list
label=['mango','mini','banana','apple','grapes','mango','apple'] 
print(label)

# load the encoder
label_encoder=preprocessing.LabelEncoder()

# Fit and trasform
op=label_encoder.fit_transform(label)
print('Encoded labels are:',op)
```

```python
#Regression
#Linear Regression  .. always use the badh data ka increase thase ka pachi decreas thase

import numpy as np 
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample data
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([[30000],[35000],[40000],[45000],[50000]])

# Train model
model=LinearRegression()
model.fit(x,y)

# Predictions
predicted_salary = model.predict([[6]])
print('The predicted salary after 6 years is:',predicted_salary)

# Vizualization
plt.scatter(x,y,color ='blue',label="Actual data")
plt.plot(x,model.predict(x),color='red',label='Regression Line')
plt.xlabel('Years of experience')            
plt.ylabel('Salary')            
plt.legend()
plt.show()   
            
```

```python
# Regresion
# Decision Tree use to if else condition
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

# Sample data
x=np.array([[1],[2],[3],[4],[5]])
# y=np.array([[30000],[35000],[40000],[45000],[50000]])
y=np.array([[30000],[15000],[40000],[25000],[50000]])

# Train model
model = DecisionTreeRegressor(max_depth=4)
model.fit(x,y)

# Predict
predicted_salary = model.predict([[6]])
print('The predicted salary after 6 years is:',predicted_salary)

# Vizualization
x_test=np.arange(1,6.1,0.01).reshape(-1,1) #1 column, no of row will be set automatically
# print(type(x_test))
# print(x_test)
y_pred=model.predict(x_test)

plt.scatter(x,y,color ='blue',label="Actual data")
plt.plot(x_test,y_pred,color='red',label='Decision Tree Regressor')
plt.xlabel('Years of experience')            
plt.ylabel('Salary')            
plt.legend()
plt.show()   
```

```python
# SVM  use both like regression and classification
# but use in regression svm called asa svr
# Regresion
# Support Vector Machine (SVM - SVR)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVR

# Input data
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([[30000],[35000],[40000],[45000],[50000]])
# y=np.array([[30000],[15000],[40000],[25000],[50000]])

# Train the SVR using RBF kernal
# rbf = Radial Basis Function kernal
# C=Regularization parameter. Higher the value,works higher to minimize the error
# epsilon = 500,error within plus/minus 500 are ignored
model =SVR(kernel = 'rbf',C=10000,epsilon=1000)

model.fit(x,y)
# Predict
print('Prediction for 6 years experience:',model.predict([[6]]))

# Vizualization
x_test=np.arange(1,6.1,0.01).reshape(-1,1)
#1 column, no of row will be set automatically
# print(type(x_test))
# print(x_test)
y_pred=model.predict(x_test)

plt.scatter(x,y,color ='blue',label="Actual data")
plt.plot(x_test,y_pred,color='red',label='SVR prediction')
plt.xlabel('Years of experience')            
plt.ylabel('Salary')            
plt.legend()
plt.show()   
```

```python
# Classification 

# 1.Naive Bayes Algorithm

import numpy as np #array mate
import pandas as pd # dataframe mate

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer



# create dataset
data = {
    'Message': [
        # Original Messages
        "Win Money now",
        "Lowest price offer",
        "Hello friend , how are you",
        "Let's meet tommorow",
        "Congratulation you won a lottery",
        "Reminder for the meeting",
        
        # New Spam Messages
        "Claim your free gift card",
        "Urgent bank account update required",
        "Exclusive deal click here now",
        "Make $5000 from home easily",
        
        # New Not_Spam Messages
        "Are we still on for lunch?",
        "Please review the attached report",
        "Happy birthday! Hope you have a great day",
        "Your package has been shipped"
    ],
    
    'Label': [
        # Original Labels
        'Spam', 'Spam', 'Not_Spam', 'Not_Spam', 'Spam', 'Not_Spam',
        
        # New Labels
        'Spam', 'Spam', 'Spam', 'Spam',
        'Not_Spam', 'Not_Spam', 'Not_Spam', 'Not_Spam'
    ]
}
df=pd.DataFrame(data)

# Map the data, convert labels (pass=1, fail=0)
df['Label']=df['Label'].map({'Not_Spam':0,'Spam':1}) #because ml algorithm can perform well on numeric data

x=df['Message']
y=df['Label']

# Convert text into numerical features
vectorizer = CountVectorizer() #converts text into numbers
x_features = vectorizer.fit_transform(x) #fit learns all unique words,
                                        # transform converts each  sentence into vector of word counts

# Train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_features,y,test_size = 0.3,random_state=1) #random state means the split must be same every time

# Train model
model = MultinomialNB()
model.fit(x_train, y_train)

# Prediction
y_pred = model.predict(x_test)

# Test with the new data

new_email = ['Free offer','Hi are we going for a movie ?']
new_features = vectorizer.transform(new_email)

prediction = model.predict(new_features)
for email,p in zip(new_email, prediction):
    print(f"Email:-->{email}-->{'Spam' if p==1 else 'Not_Spam'}")
    
```
