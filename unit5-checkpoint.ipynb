```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
img=cv2.imread('i1.jpg')#image read 
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
#KERNAL
# diension koi pan ai sako
k1=np.array([[3,6,0],[0,5,0],[0,0,7]])
#syntax:
# cv2.filter2D(source_image,ddepth,name_of_kernal)
k1output=cv2.filter2D(img,-1,k1)
#(row,col,posi) imshow means image show
plt.subplot(1,2,1),plt.imshow(img),plt.title('source')
plt.subplot(1,2,2),plt.imshow(k1output),plt.axis('off') # thi axis no ave
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
img=cv2.imread('i1.jpg')#image read 
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
#KERNAL
# diension koi pan ai sako
k1=np.array([[0,0,0],[0,2,0],[0,0,0]])
k2=np.array([[-1,-1,-1],[-1,8,-1],[-1,1,-1]])
#syntax:
# cv2.filter2D(source_image,ddepth,name_of_kernal)
k1output=cv2.filter2D(img,-1,k1)
k2output=cv2.filter2D(img,-1,k2)
#(row,col,posi) imshow means image show
plt.subplot(1,3,1),plt.imshow(img),plt.title('source')
plt.subplot(1,3,2),plt.imshow(k1output),plt.axis('off') # thi axis no ave
plt.subplot(1,3,3),plt.imshow(k2output),plt.axis('off') # thi axis no ave
```

```python
# blurring image
import cv2
import matplotlib.pyplot as plt
import numpy as np
img=cv2.imread('i1.jpg')#image read 
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
# ones means 5*5 badha metrx ma 1 ave
k1=np.ones((5,5),np.float32)/16
k1op=cv2.filter2D(img,-1,k1)
plt.subplot(1,2,1),plt.imshow(img),plt.axis('off'),plt.title('source')
plt.subplot(1,2,2),plt.imshow(k1op),plt.axis('off'),plt.title('k1op')
```

```python
# differrent types of blurring
import cv2
import matplotlib.pyplot as plt
import numpy as np
img=cv2.imread('i1.jpg')#image read 
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
# 1.box blur
box=cv2.blur(img,(5,5))
# gaussian blur
gauss=cv2.GaussianBlur(img,(9,9),0)#dimensions of the kernal must be of any odd no
plt.subplot(2,2,1),plt.imshow(img),plt.axis('off'),plt.title('source')
plt.subplot(2,2,2),plt.imshow(box),plt.axis('off'),plt.title('box')
plt.subplot(2,2,3),plt.imshow(gauss),plt.axis('off'),plt.title('box')
```
