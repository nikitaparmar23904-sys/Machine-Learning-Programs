```python
# unit-5
import matplotlib.pyplot as plt
import cv2
import numpy as np
img=cv2.imread('i1.jpg')
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
k1=np.array([[0,0,0],[0,8,0],[0,0,0]])
op=cv2.filter2D(img,-1,k1)
plt.subplot(1,2,1),plt.imshow(img),plt.title('source image')
plt.subplot(1,2,2),plt.imshow(op),plt.title('source image')
```

```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
img=cv2.imread('i1.jpg')
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
box=cv2.blur(img,(3,3))
g=cv2.GaussianBlur(img,(31,3),0)
plt.subplot(2,2,1),plt.imshow(img),plt.title('source image')
plt.subplot(2,2,2),plt.imshow(box),plt.title('source image')
plt.subplot(2,2,3),plt.imshow(g),plt.title('source image')
```
